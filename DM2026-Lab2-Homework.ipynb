## 1. Model Development (10 pts Required)

This report describes our approach to the 6-class emotion classification task (anger, disgust, fear, joy, sadness, and surprise) on short social-media posts. The dataset consists of 3,360 labeled training posts and 1,440 test posts, with a median length of 13 words. In this section, we first summarize the submitted models and their scores. Subsequently, Sec. 1.1 and Sec. 1.2 introduce the preprocessing and feature engineering steps, and Sec. 1.3 explains the final model. Finally, Sec. 1.4 summarizes how AI assistance was used during development and why each suggestion was adopted or rejected.

Our final model is a weighted soft-voting ensemble of three Logistic Regression experts. Specifically, the experts are trained on EmbeddingGemma-2 (270M) embeddings, Qwen3-Embedding-8B embeddings, and TF-IDF word and character n-grams, respectively, and their predicted probabilities are combined with weights of 1, 2, and 1. Additionally, an exact-match rule is applied to the test posts that appear verbatim in the training set. This model achieves the highest cross-validation (CV) and Public Leaderboard scores, and it is the model explained in Sec. 1.3.

However, the submission counted on the Private Leaderboard (rank 66, as shown in the screenshot above) is an earlier version, namely the equal-weight ensemble of Gemma-2 and Qwen3 with the same exact-match rule. ### 1.1 Preprocessing Steps

In this section, we first describe how the competition data are loaded. Subsequently, the data exploration that shaped our pipeline is presented. Finally, we discuss the text cleaning and sample filtering methods, which were tested but not adopted in the final pipeline.

#### 1) Data Loading

The competition data are provided in three files, namely `final_posts.json` (post text), `data_identification.csv` (train/test split), and `emotion.csv` (labels of the training posts only). We join the three files on `post_id` to obtain the training and test sets.

#### 2) Data Exploration

Before designing the model, we conducted five checks on the data. Firstly, the class distribution is imbalanced: joy has 875 posts, anger 674, sadness 628, disgust 454, surprise 441, and fear 288, so the largest class is about three times the smallest. We tried `class_weight="balanced"` and oversampling to compensate for this imbalance. However, neither of them improved the CV score (0.604 and 0.592, compared with 0.602), and therefore the natural distribution is kept.

Secondly, we applied the Pearson-correlation keyword features from Lab 1, taking the top 10 keywords of each class. We found that 61.8% of the posts contain none of these keywords, and the best per-class Pearson r is only 0.14 to 0.31. This explains why the TF-IDF approach in the Master notebook, which reaches 0.86 on the Lab 2 tweets, drops to 0.46 on this dataset. It also motivates the use of dense sentence embeddings instead of keyword features.

Thirdly, the training set contains 50 duplicated texts, and 16 of them (32%) have conflicting labels. This indicates label noise, which limits the achievable F1 score. Fourthly, 59 test posts are identical to a training post, and these posts are handled by the exact-match rule in Sec. 1.3. Finally, adversarial validation between the training and test sets gives an AUC of 0.510, and the distributions of prediction entropy on the two sets are almost identical, as shown in Fig. 1. This suggests that there is no distribution shift between them, so the CV score is a reliable estimate of the leaderboard score.

![Prediction entropy of the training and test posts](./pics/entropy_train_vs_test.png)

**Fig. 1:** Distribution of the normalized prediction entropy of Gemma-2 + LR on the training posts (out-of-fold) and the test posts (model fitted on all training posts). The two distributions are almost identical.

#### 3) Text Cleaning

We tested several text cleaning operations, including the removal of placeholders (`[NAME]`, `@<username>`), stop words, emoji, curly quotes, repeated letters and punctuation, all-caps words, and extra whitespace. However, none of them improved the performance of the large embedding models. Specifically, the ensemble scores 0.617 with full cleaning and 0.616 without it. Furthermore, stop-word removal decreases the CV score of Gemma-2 by 0.03, because it breaks phrases such as "fuck off" into "fuck". Hence, the raw text is fed to the embedding models, and the only normalization is lower-casing for the TF-IDF expert.

#### 4) Sample Filtering

Since the labels are noisy, we tried several methods to remove or relabel unrepresentative training posts, as shown in Table 2.

Therefore, we keep the original labels.

### 1.2 Feature Engineering Steps

In this section, we introduce the three feature representations used in the final model. Each representation serves as the input of a separate classifier, referred to as an expert. Subsequently, we compare the candidate encoders evaluated before choosing these three. Finally, we list the features that were tested but not adopted.

#### 1) EmbeddingGemma-2 Sentence Embeddings

Each post is encoded into a 768-dimensional vector by EmbeddingGemma-2 (270M), which is served locally through Ollama. Following the task prompt designed for this model, we prepend `"task: classification | query: "` to each text before encoding. This prefix alone raises the CV score from 0.583 to 0.602.

#### 2) Qwen3-Embedding-8B Sentence Embeddings

Each post is encoded into a 4,096-dimensional vector by Qwen3-Embedding-8B. Specifically, each text is wrapped in the instruction format given in the model card, as shown below,

```
Instruct: Classify the emotion expressed in the given post as one of anger, disgust, fear, joy, sadness, surprise
Query: <text>
```

where `<text>` represents the post. This instruction raises the CV score from 0.575 to 0.605.

#### 3) TF-IDF n-grams

The third representation is a sparse TF-IDF vector, which concatenates word 1–2-grams (`min_df=2`) with character 2–5-grams (`analyzer="char_wb"`, `min_df=3`). Both vectorizers use `sublinear_tf=True` and are applied to lower-cased text. To avoid data leakage, the vectorizers are fitted inside each CV fold. On its own, this expert is weak (0.491). However, it captures word choice, whereas the embeddings capture meaning, so the errors of the two types of experts differ. The same observation was made in Sec. 8 of the Master notebook.

#### 4) Comparison of Encoders

Before choosing the above representations, we compared several encoders, as shown in Table 3.

From the experimental results, we observe that switching from MiniLM to EmbeddingGemma-2 brings the largest single improvement, about 0.09. Furthermore, Qwen3-Embedding-8B is 28 times larger than EmbeddingGemma-2, yet it is only 0.003 better on its own. This indicates that the gain from a larger encoder diminishes quickly with model size.

#### 5) Features Not Adopted

We also tested the NRC Emotion Lexicon counts as features. They score only 0.283 on their own and bring no gain when appended to the Gemma-2 embeddings. Additionally, reducing the Qwen3 embeddings to 256 dimensions with PCA gives the same score and only speeds up training. Therefore, neither of them is included in the final model.

### 1.3 Explanation of Your Model

In this section, we first provide an overview of the final model, as illustrated in Fig. 2, and describe its three components, namely the experts, the soft voting, and the exact-match rule. Subsequently, we explain why a linear classifier and an ensemble are adopted, and we discuss the approaches that did not make the final model. Finally, an error analysis is presented to explain the performance ceiling.

**Fig. 2:** Pipeline of the final model. P₁, P₂, and P₃ represent the probabilities of the 6 emotions predicted by each expert, and ×1 and ×2 represent the voting weights.

```
post text ─┬─ EmbeddingGemma-2 (prefix) ──── LogisticRegression ───────── P₁ ─┐ ×1
           ├─ Qwen3-Embedding-8B (instr.) ── LogisticRegression ───────── P₂ ─┤ ×2  →  argmax  →  label
           └─ TF-IDF word+char n-grams ───── LogisticRegression (C=4) ─── P₃ ─┘ ×1
                                                         exact train match? → use train label
```

#### 1) Experts

Each expert is a multinomial `LogisticRegression(max_iter=3000)`, which outputs a probability for each of the 6 emotions. The regularization parameter C is set to 1 for the two embedding experts and 4 for the TF-IDF expert.

#### 2) Soft Voting

The final probability is calculated as the weighted sum of the expert outputs, as shown in Eq. 1,

$$P = 1 \cdot P_1 + 2 \cdot P_2 + 1 \cdot P_3 \tag{1}$$

where P₁, P₂, and P₃ represent the probabilities predicted by the Gemma-2, Qwen3, and TF-IDF experts, respectively. The predicted label is the emotion with the highest value in P. The weights were chosen by a grid search over {0, 0.5, 1, 2} for each expert. Specifically, the search is scored on the out-of-fold probabilities, so no expert is evaluated on the posts it was trained on. The candidate experts in the search were Gemma-2, Qwen3, TF-IDF, a RAG-style LLM classifier, and the lexicon model, and the search assigned a weight of 0 to the last two.

#### 3) Exact-Match Rule

For the 59 test posts whose text appears verbatim in the training set, the model prediction is replaced with the training label. If the text appears more than once in the training set, the majority label is used.

#### 4) Why a Linear Classifier

We compared several classifier heads on the Gemma-2 embeddings, including Logistic Regression with tuned C and class weights (best 0.604), a PyTorch softmax layer with a weight decay sweep and a warm-up plus cosine annealing schedule, and a 2-layer MLP with 256 to 1,024 hidden units and a dropout rate of 0.3 to 0.5. However, the validation loss always reached its minimum at about 1.02, compared with ln 6 ≈ 1.79 for random guessing, and the MLP only overfitted faster, as shown in Fig. 3. This indicates that the bottleneck lies in the information contained in the embeddings rather than in the classifier. Therefore, we keep Logistic Regression as the simplest and most stable head.

![Training curves of the linear head and the MLPs](./pics/torch_mlp.png)

**Fig. 3:** Training curves of the linear softmax layer, denoted as linear (ref), and 2-layer MLPs on the Gemma-2 embeddings, where h, drop, and wd represent the hidden size, dropout rate, and weight decay, respectively. The left plot shows the validation loss (solid) and the training loss (dotted), and the right plot shows the validation macro F1.

On the other hand, clustering-based alternatives performed worse. K-means followed by cluster labeling reached 0.552 at k = 48, while DBSCAN failed with at most 0.185. Furthermore, the t-SNE visualization in Fig. 4 shows that each emotion forms a tight core, with the remaining points scattered across the whole map. This suggests that emotions correspond to directions in the embedding space rather than separate clusters, which favors a linear classifier.

![t-SNE of the EmbeddingGemma-2 embeddings](./pics/tsne_gemma2_train.png)

**Fig. 4:** t-SNE visualization of the EmbeddingGemma-2 embeddings of the 3,360 training posts. The left plot shows all emotions, and the right plots highlight each emotion separately.

#### 5) Why an Ensemble

The improvements came only from combining models whose errors differ, as shown in Table 4.

**Table 4:** CV performance of expert combinations. All combinations use equal weights unless otherwise noted. NaN represents that the score was not recorded.

| Combination | CV macro F1 | CV weighted F1 |
|---|---|---|
| Gemma-2 alone | 0.602 | 0.618 |
| Gemma-2 + Qwen3 | 0.616 | 0.635 |
| Gemma-2 + Qwen3 + mE5 | 0.611 | NaN |
| Gemma-2 + Qwen3 + MiniLM | 0.607 | NaN |
| Stacking (meta Logistic Regression over 4 experts) | 0.613 | NaN |
| **Gemma-2 ×1 + Qwen3 ×2 + TF-IDF ×1** | **0.627** | **0.646** |

From the experimental results, we observe that adding a weaker embedding expert such as mE5 or MiniLM lowers the score, whereas adding the TF-IDF expert raises it. This is attributed to the fact that the TF-IDF expert makes different errors from the embedding experts. Since the weights were selected on the same CV, the score of 0.627 is slightly optimistic. Nevertheless, the top weight settings all lie between 0.624 and 0.627, which is consistently about 0.01 above the equal-weight ensemble.

#### 6) Approaches Not Adopted

We also tried two more complex approaches. Firstly, a RAG-style LLM classifier retrieves the 10 most similar labeled training posts and asks a local 8B LLM (gemma4) to classify the post. It scores 0.568 on its own and brings no gain to the ensemble. Secondly, we fine-tuned DeBERTa-v3 (large and base) with group 5-fold CV and several learning rates. The best setting (base, learning rate 3e-5) reaches 0.588, but adding it to the ensemble lowers the score to 0.623. Therefore, neither approach is included in the final model.

#### 7) Error Analysis

To explain why the performance ceiling is about 0.62, we computed two values for each training post from the out-of-fold probabilities of the Gemma-2 + LR expert, namely the prediction entropy, which measures how uncertain the model is, and the cross-entropy (CE) with the label, which measures how much the model disagrees with the label. Specifically, the entropy is normalized by ln 6 so that it ranges from 0 to 1. Based on an entropy threshold of 0.8 and a CE threshold of ln 6 ≈ 1.79, which is the CE of a uniform guess, the posts are divided into four regions, as shown in Fig. 5 and Table 5.

The main confusions are disgust and anger, surprise predicted as joy, and sadness predicted as joy.

### 1.4 AI Prompts, what and why you decide to use that technique that suggest by AI

In this section, we summarize how AI assistance was used during development and why each suggestion was adopted or rejected. An AI assistant was consulted for experiment ideas and implementation throughout the competition. However, no suggestion was adopted directly. Specifically, every suggestion was first implemented as an experiment and then accepted only if it raised the 5-fold out-of-fold macro F1 on the shared split described at the beginning of this report. Additionally, a local LLM (gemma4 via Ollama) was used as a component in two experiments, namely the HyDE anchors and the RAG-style classifier.

The adopted suggestions are those that passed this criterion. Firstly, replacing MiniLM with stronger pre-trained encoders (EmbeddingGemma-2 and Qwen3-Embedding-8B) raised the score by about 0.09. Secondly, combining models with different error types in an ensemble raised it by about 0.025, and adding a TF-IDF expert to the ensemble contributed 0.011 of this gain. Finally, the analysis of the overlap between the training and test posts led to the exact-match rule.

On the other hand, most suggestions were rejected because they did not raise the score. These include more expressive classifier heads (Sec. 1.3), clustering-based classification, sample filtering and soft relabeling (Sec. 1.1), the RAG-style classifier, the emotion lexicon, oversampling, and transformer fine-tuning. In particular, every method that removed or rewrote training labels failed, because the test set contains the same proportion of ambiguous posts as the training set.

Furthermore, one suggestion was rejected for a different reason. While surveying strategies for unrepresentative samples, the AI assistant found that many posts overlap with the public GoEmotions dataset, and it later noted that common ready-made emotion models are trained on this dataset. We used neither the dataset nor such models, because doing so would leak the test labels. Therefore, the final model consists only of general-purpose pre-trained embeddings, a TF-IDF expert, and a weighted ensemble.

````python
### Add the code related to the preprocessing steps in cells inside this section
# Report Sec. 1.1: load the competition files, run the data checks, and fix the 5-fold split.
import json
from pathlib import Path

import numpy as np
import pandas as pd

DATA = Path("data/kaggle")  # competition files and cached embeddings (not pushed to the repo)


def load_data():
    """Join posts, split and labels on post_id; test rows keep label = NaN."""
    posts = pd.DataFrame(json.load(open(DATA / "final_posts.json", encoding="utf-8")))
    split = pd.read_csv(DATA / "data_identification.csv")
    labels = pd.read_csv(DATA / "emotion.csv")
    df = posts.merge(split, on="post_id").merge(labels, on="post_id", how="left")
    train_df = df[df["set"] == "train"].reset_index(drop=True)
    test_df = df[df["set"] == "test"].reset_index(drop=True)
    return train_df, test_df


tr, te = load_data()
y = tr["label"].to_numpy()
classes = np.array(sorted(set(y)))
print(f"train {len(tr)} | test {len(te)} | median words {tr['text'].str.split().str.len().median():.0f}")

# Data checks (Sec. 1.1, 2) Data Exploration)
print("class counts:", tr["label"].value_counts().to_dict())
dup = tr.groupby("text")["label"].agg(["size", "nunique"]).query("size > 1")
print(f"duplicated train texts {len(dup)}, with conflicting labels {(dup['nunique'] > 1).sum()}")
print(f"test posts identical to a train post {te['text'].isin(set(tr['text'])).sum()}")

# Text cleaning was tested and not adopted: the embedding models get the raw text,
# and the TF-IDF expert only gets lower-cased text.
txt_tr, txt_te = tr["text"].str.lower().to_numpy(), te["text"].str.lower().to_numpy()

# One fixed split shared by every experiment, so all CV scores are comparable.
from sklearn.model_selection import StratifiedKFold
FOLDS = list(StratifiedKFold(5, shuffle=True, random_state=42).split(np.zeros(len(y)), y))
````

## 2. Feature Engineering Steps

````python
### Add the code related to the feature engineering steps in cells inside this section
# Report Sec. 1.2: two sentence-embedding representations and one TF-IDF representation.
import requests
from scipy.sparse import hstack
from sklearn.feature_extraction.text import TfidfVectorizer

OLLAMA = "http://127.0.0.1:11435/api/embed"  # local Ollama server that hosts the embedding models


def embed(texts, cache_name, model, prefix, batch=32):
    """Encode texts with an Ollama embedding model; reuse the cached .npy if it exists."""
    cache = DATA / f"{cache_name}.npy"
    if cache.exists():
        return np.load(cache)
    out = []
    for i in range(0, len(texts), batch):
        r = requests.post(OLLAMA, json={"model": model, "input": [prefix + t for t in texts[i:i + batch]]}, timeout=900)
        r.raise_for_status()
        out.extend(r.json()["embeddings"])
    v = np.array(out, dtype=np.float32)
    np.save(cache, v)
    return v


# 1) EmbeddingGemma-2 (270M), 768 dimensions, with the task prompt for classification.
GEMMA, GEMMA_PREFIX = "embeddinggemma-2:270m", "task: classification | query: "
X_gemma = embed(tr["text"].tolist(), "gemma2_kaggle_train_cls", GEMMA, GEMMA_PREFIX)
X_gemma_te = embed(te["text"].tolist(), "gemma2_kaggle_test_cls", GEMMA, GEMMA_PREFIX)

# 2) Qwen3-Embedding-8B, 4,096 dimensions, with the instruction format from the model card.
QWEN = "qwen3-embedding:8b-fp16"
QWEN_PREFIX = ("Instruct: Classify the emotion expressed in the given post as one of "
               "anger, disgust, fear, joy, sadness, surprise\nQuery: ")
X_qwen = embed(tr["text"].tolist(), "qwen3_8b_kaggle_train_instruct", QWEN, QWEN_PREFIX)
X_qwen_te = embed(te["text"].tolist(), "qwen3_8b_kaggle_test_instruct", QWEN, QWEN_PREFIX)
print("Gemma-2", X_gemma.shape, "| Qwen3", X_qwen.shape)


# 3) TF-IDF word 1-2-grams + character 2-5-grams. The vectorizers are fitted on the
#    training part only, so inside CV they never see the validation fold.
def tfidf_feats(fit_texts, texts_list):
    w = TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True).fit(fit_texts)
    c = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=3, sublinear_tf=True).fit(fit_texts)
    return [hstack([w.transform(t), c.transform(t)]).tocsr() for t in texts_list]
````

## 3. Model Implementation Steps

````python
### Add the code related to the model implementation steps in cells inside this section
# Report Sec. 1.3: three Logistic Regression experts, weighted soft voting, and the exact-match rule.
import itertools

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score


def macro(p):
    return f1_score(y, classes[np.asarray(p).argmax(1)], average="macro")


def lr_probs(X, X_te, C=1.0):
    """Out-of-fold probabilities on the train set, and test probabilities from a fit on all train posts."""
    oof = np.zeros((len(y), len(classes)))
    for fi, hi in FOLDS:
        oof[hi] = LogisticRegression(max_iter=3000, C=C).fit(X[fi], y[fi]).predict_proba(X[hi])
    return oof, LogisticRegression(max_iter=3000, C=C).fit(X, y).predict_proba(X_te)


# 1) Experts
experts = {"gemma2": lr_probs(X_gemma, X_gemma_te), "qwen3": lr_probs(X_qwen, X_qwen_te)}
oof = np.zeros((len(y), len(classes)))
for fi, hi in FOLDS:  # TF-IDF is refitted inside each fold
    A, B = tfidf_feats(txt_tr[fi], [txt_tr[fi], txt_tr[hi]])
    oof[hi] = LogisticRegression(max_iter=3000, C=4).fit(A, y[fi]).predict_proba(B)
A, B = tfidf_feats(txt_tr, [txt_tr, txt_te])
experts["tfidf"] = (oof, LogisticRegression(max_iter=3000, C=4).fit(A, y).predict_proba(B))
for k, (o, _) in experts.items():
    print(f"{k:7s} CV macro F1 {macro(o):.3f}")

# 2) Soft voting: grid search of the weights, scored on out-of-fold probabilities only.
names = list(experts)
scores = []
for w in itertools.product([0, 0.5, 1, 2], repeat=len(names)):
    if sum(w) == 0:
        continue
    p = sum(wi * experts[k][0] for wi, k in zip(w, names))
    scores.append((macro(p), f1_score(y, classes[p.argmax(1)], average="weighted"), w))
scores.sort(reverse=True)
best_w = scores[0][2]
print("best weights", dict(zip(names, best_w)), f"CV macro F1 {scores[0][0]:.3f}")

# 3) Exact-match rule: a test post identical to a train post takes that post's (majority) label.
train_label = tr.groupby("text")["label"].agg(lambda s: s.mode().iloc[0])
exact = te["text"].map(train_label)


def write_submission(weights, name):
    P = sum(wi * experts[k][1] for wi, k in zip(weights, names))
    pred = pd.Series(classes[P.argmax(1)], index=te.index).where(exact.isna(), exact)
    sub = pd.DataFrame({"id": te["post_id"], "label": pred})
    sample = pd.read_csv(DATA / "sampleSubmission.csv")  # keep Kaggle's row order
    sub = sample[["post_id"]].rename(columns={"post_id": "id"}).merge(sub, on="id", how="left")
    assert sub["label"].notna().all()
    sub.to_csv(DATA / name, index=False)
    print("wrote", name, sub["label"].value_counts().to_dict())


write_submission(best_w, "submission_notebook_final.csv")         # final model (Gemma-2 x1 + Qwen3 x2 + TF-IDF x1)
write_submission((1, 1, 0), "submission_notebook_scored.csv")     # scored submission (equal-weight Gemma-2 + Qwen3)
````
